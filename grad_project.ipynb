```python
# CEFR Level Classification using BERT - Complete Training Pipeline
# Optimized for Google Colab with comprehensive evaluation and overfitting prevention

# ============================================================================
# SECTION 1: Setup and Installation
# ============================================================================

!pip install -q transformers datasets scikit-learn matplotlib seaborn plotly torch accelerate

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    EarlyStoppingCallback
)
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score,
    precision_recall_fscore_support
)
from sklearn.utils.class_weight import compute_class_weight
import warnings
warnings.filterwarnings('ignore')

# Set random seeds for reproducibility
np.random.seed(42)
torch.manual_seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(42)

# Check GPU availability
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"Memory Available: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")

# ============================================================================
# SECTION 2: Data Loading and Exploration
# ============================================================================

print("\n" + "="*80)
print("LOADING AND EXPLORING DATASET")
print("="*80)

# Upload the dataset file
from google.colab import files
import os

# Check if file already exists
if not os.path.exists('final_combined_cefr_english_dataset.csv'):
    print("\n📁 Please upload your CSV file:")
    uploaded = files.upload()

    # Get the uploaded filename
    uploaded_filename = list(uploaded.keys())[0]

    # Rename to expected filename if different
    if uploaded_filename != 'final_combined_cefr_english_dataset.csv':
        os.rename(uploaded_filename, 'final_combined_cefr_english_dataset.csv')
        print(f"✓ File renamed to: final_combined_cefr_english_dataset.csv")
else:
    print("✓ Dataset file found!")

# Load the dataset
df = pd.read_csv('final_combined_cefr_english_dataset.csv')

print(f"\nDataset Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print("\nFirst few rows:")
print(df.head())

# Basic statistics
print("\n" + "-"*80)
print("DATASET STATISTICS")
print("-"*80)
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())

# Identify text and label columns (adjust these based on your actual column names)
# Common column names: 'text', 'essay', 'writing', 'content' for text
# 'label', 'level', 'cefr', 'proficiency' for labels
text_col = None
label_col = None

for col in df.columns:
    col_lower = col.lower()
    if any(keyword in col_lower for keyword in ['text', 'essay', 'writing', 'content', 'sentence']):
        text_col = col
    if any(keyword in col_lower for keyword in ['label', 'level', 'cefr', 'proficiency', 'class']):
        label_col = col

if text_col is None or label_col is None:
    print("\n⚠️  Could not automatically detect columns. Please specify:")
    print(f"Available columns: {df.columns.tolist()}")
    text_col = input("Enter the name of the TEXT column: ")
    label_col = input("Enter the name of the LABEL column: ")

print(f"\n✓ Using columns:")
print(f"  Text column: '{text_col}'")
print(f"  Label column: '{label_col}'")

# Clean the data
df = df[[text_col, label_col]].dropna()
df[text_col] = df[text_col].astype(str)

# Text length statistics
df['text_length'] = df[text_col].str.len()
df['word_count'] = df[text_col].str.split().str.len()

print("\n" + "-"*80)
print("TEXT LENGTH STATISTICS")
print("-"*80)
print(df[['text_length', 'word_count']].describe())

# ============================================================================
# SECTION 3: Class Distribution Analysis
# ============================================================================

print("\n" + "="*80)
print("CLASS DISTRIBUTION ANALYSIS")
print("="*80)

# Class distribution
class_dist = df[label_col].value_counts().sort_index()
print("\nClass Distribution:")
print(class_dist)
print(f"\nTotal samples: {len(df)}")

# Check for class balance
balance_ratio = class_dist.min() / class_dist.max()
print(f"\nBalance Ratio (min/max): {balance_ratio:.3f}")
if balance_ratio < 0.5:
    print("⚠️  WARNING: Significant class imbalance detected!")
    print("   → Will use class weights and stratified sampling")
else:
    print("✓ Classes are reasonably balanced")

# Visualize class distribution
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Class Distribution (Count)', 'Class Distribution (%)',
                    'Text Length by Class', 'Word Count by Class'),
    specs=[[{'type': 'bar'}, {'type': 'pie'}],
           [{'type': 'box'}, {'type': 'box'}]]
)

# Bar chart
fig.add_trace(
    go.Bar(x=class_dist.index, y=class_dist.values, name='Count',
           marker_color='steelblue'),
    row=1, col=1
)

# Pie chart
fig.add_trace(
    go.Pie(labels=class_dist.index, values=class_dist.values, name='Distribution'),
    row=1, col=2
)

# Box plot for text length
for level in sorted(df[label_col].unique()):
    data = df[df[label_col] == level]['text_length']
    fig.add_trace(
        go.Box(y=data, name=str(level)),
        row=2, col=1
    )

# Box plot for word count
for level in sorted(df[label_col].unique()):
    data = df[df[label_col] == level]['word_count']
    fig.add_trace(
        go.Box(y=data, name=str(level), showlegend=False),
        row=2, col=2
    )

fig.update_layout(height=800, showlegend=True, title_text="Dataset Analysis Dashboard")
fig.show()

# ============================================================================
# SECTION 4: Data Preprocessing and Splitting
# ============================================================================

print("\n" + "="*80)
print("DATA PREPROCESSING")
print("="*80)

# Encode labels
label_mapping = {label: idx for idx, label in enumerate(sorted(df[label_col].unique()))}
reverse_mapping = {idx: label for label, idx in label_mapping.items()}
df['label_encoded'] = df[label_col].map(label_mapping)

print("\nLabel Mapping:")
for label, idx in label_mapping.items():
    print(f"  {label} → {idx}")

# Stratified split: 70% train, 15% validation, 15% test
train_df, temp_df = train_test_split(
    df, test_size=0.3, random_state=42, stratify=df['label_encoded']
)
val_df, test_df = train_test_split(
    temp_df, test_size=0.5, random_state=42, stratify=temp_df['label_encoded']
)

print(f"\nDataset Split:")
print(f"  Training:   {len(train_df)} samples ({len(train_df)/len(df)*100:.1f}%)")
print(f"  Validation: {len(val_df)} samples ({len(val_df)/len(df)*100:.1f}%)")
print(f"  Test:       {len(test_df)} samples ({len(test_df)/len(df)*100:.1f}%)")

# Verify stratification
print("\nClass distribution in each set:")
for name, subset in [('Train', train_df), ('Val', val_df), ('Test', test_df)]:
    dist = subset[label_col].value_counts(normalize=True).sort_index()
    print(f"\n{name}:")
    print(dist.to_string())

# ============================================================================
# SECTION 5: Model Setup
# ============================================================================

print("\n" + "="*80)
print("MODEL SETUP")
print("="*80)

# Model selection
MODEL_NAME = "distilbert-base-uncased"  # Faster and lighter than BERT
# Alternative: "bert-base-uncased" for potentially better accuracy

print(f"\nSelected Model: {MODEL_NAME}")

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Tokenization function
MAX_LENGTH = 512

def tokenize_function(examples):
    return tokenizer(
        examples,
        padding='max_length',
        truncation=True,
        max_length=MAX_LENGTH,
        return_tensors='pt'
    )

# Custom Dataset class
class CEFRDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_length):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts.iloc[idx])
        label = self.labels.iloc[idx]

        encoding = self.tokenizer(
            text,
            padding='max_length',
            truncation=True,
            max_length=self.max_length,
            return_tensors='pt'
        )

        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': torch.tensor(label, dtype=torch.long)
        }

# Create datasets
train_dataset = CEFRDataset(
    train_df[text_col].reset_index(drop=True),
    train_df['label_encoded'].reset_index(drop=True),
    tokenizer,
    MAX_LENGTH
)

val_dataset = CEFRDataset(
    val_df[text_col].reset_index(drop=True),
    val_df['label_encoded'].reset_index(drop=True),
    tokenizer,
    MAX_LENGTH
)

test_dataset = CEFRDataset(
    test_df[text_col].reset_index(drop=True),
    test_df['label_encoded'].reset_index(drop=True),
    tokenizer,
    MAX_LENGTH
)

print(f"\n✓ Datasets created:")
print(f"  Train: {len(train_dataset)} samples")
print(f"  Val:   {len(val_dataset)} samples")
print(f"  Test:  {len(test_dataset)} samples")

# Compute class weights for imbalanced data
class_weights = compute_class_weight(
    'balanced',
    classes=np.unique(train_df['label_encoded']),
    y=train_df['label_encoded']
)
class_weights = torch.tensor(class_weights, dtype=torch.float).to(device)
print(f"\nClass weights: {class_weights}")

# ============================================================================
# SECTION 6: Training Configuration with Overfitting Prevention
# ============================================================================

print("\n" + "="*80)
print("TRAINING CONFIGURATION")
print("="*80)

# Load model
num_labels = len(label_mapping)
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=num_labels
)
model.to(device)

print(f"\nModel loaded with {num_labels} output classes")
print(f"Total parameters: {sum(p.numel() for p in model.parameters()):,}")
print(f"Trainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")

# Training arguments with comprehensive overfitting prevention
training_args = TrainingArguments(
    output_dir='./results',
    num_train_epochs=10,  # More epochs with early stopping
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    learning_rate=2e-5,
    weight_decay=0.01,  # L2 regularization
    warmup_steps=500,
    logging_dir='./logs',
    logging_steps=50,
    eval_strategy='steps',
    eval_steps=100,
    save_strategy='steps',
    save_steps=100,
    load_best_model_at_end=True,
    metric_for_best_model='f1',
    greater_is_better=True,
    save_total_limit=3,
    seed=42,
    fp16=torch.cuda.is_available(),  # Mixed precision training
    dataloader_num_workers=2,
    gradient_accumulation_steps=2,  # Effective batch size = 32
    max_grad_norm=1.0,  # Gradient clipping
    report_to='none',  # Disable W&B and other logging platforms
)

print("\n✓ Training configuration:")
print(f"  Epochs: {training_args.num_train_epochs}")
print(f"  Batch size: {training_args.per_device_train_batch_size}")
print(f"  Learning rate: {training_args.learning_rate}")
print(f"  Weight decay: {training_args.weight_decay}")
print(f"  Early stopping: Enabled (patience=3)")

# Custom Trainer with class weights
class WeightedTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits = outputs.logits
        loss_fct = torch.nn.CrossEntropyLoss(weight=class_weights)
        loss = loss_fct(logits.view(-1, self.model.config.num_labels), labels.view(-1))
        return (loss, outputs) if return_outputs else loss

# Evaluation metrics
def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = np.argmax(predictions, axis=1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels, predictions, average='weighted'
    )
    acc = accuracy_score(labels, predictions)

    return {
        'accuracy': acc,
        'f1': f1,
        'precision': precision,
        'recall': recall
    }

# Initialize trainer with early stopping
trainer = WeightedTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=3)]
)

# ============================================================================
# SECTION 7: Model Training
# ============================================================================

print("\n" + "="*80)
print("STARTING TRAINING")
print("="*80)

# Train the model
train_result = trainer.train()

print("\n✓ Training completed!")
print(f"  Training time: {train_result.metrics['train_runtime']:.2f} seconds")
print(f"  Training loss: {train_result.metrics['train_loss']:.4f}")

# Save the model
trainer.save_model('./best_model')
tokenizer.save_pretrained('./best_model')
print("\n✓ Model saved to './best_model'")

# ============================================================================
# SECTION 8: Training History Visualization
# ============================================================================

print("\n" + "="*80)
print("VISUALIZING TRAINING HISTORY")
print("="*80)

# Extract training history
log_history = trainer.state.log_history

train_loss = [log['loss'] for log in log_history if 'loss' in log]
eval_loss = [log['eval_loss'] for log in log_history if 'eval_loss' in log]
eval_acc = [log['eval_accuracy'] for log in log_history if 'eval_accuracy' in log]
eval_f1 = [log['eval_f1'] for log in log_history if 'eval_f1' in log]

# Create training visualization
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Training & Validation Loss', 'Validation Accuracy',
                    'Validation F1 Score', 'Loss Comparison'),
    specs=[[{'secondary_y': False}, {'secondary_y': False}],
           [{'secondary_y': False}, {'secondary_y': False}]]
)

# Plot 1: Training loss
steps_train = list(range(len(train_loss)))
fig.add_trace(
    go.Scatter(x=steps_train, y=train_loss, name='Train Loss',
               line=dict(color='blue')),
    row=1, col=1
)

# Plot 2: Validation loss
steps_eval = list(range(len(eval_loss)))
fig.add_trace(
    go.Scatter(x=steps_eval, y=eval_loss, name='Val Loss',
               line=dict(color='red')),
    row=1, col=1
)

# Plot 3: Validation accuracy
fig.add_trace(
    go.Scatter(x=steps_eval, y=eval_acc, name='Val Accuracy',
               line=dict(color='green'), showlegend=False),
    row=1, col=2
)

# Plot 4: Validation F1
fig.add_trace(
    go.Scatter(x=steps_eval, y=eval_f1, name='Val F1',
               line=dict(color='purple'), showlegend=False),
    row=2, col=1
)

# Plot 5: Loss gap (overfitting indicator)
min_len = min(len(train_loss), len(eval_loss))
loss_gap = [train_loss[i] - eval_loss[i] for i in range(min_len)]
fig.add_trace(
    go.Scatter(x=list(range(min_len)), y=loss_gap, name='Loss Gap',
               line=dict(color='orange'), fill='tozeroy', showlegend=False),
    row=2, col=2
)

fig.update_layout(height=800, title_text="Training Metrics Dashboard")
fig.show()

# Check for overfitting
final_train_loss = train_loss[-1]
final_val_loss = eval_loss[-1]
loss_gap_final = final_val_loss - final_train_loss

print("\n" + "-"*80)
print("OVERFITTING ANALYSIS")
print("-"*80)
print(f"Final Training Loss: {final_train_loss:.4f}")
print(f"Final Validation Loss: {final_val_loss:.4f}")
print(f"Loss Gap: {loss_gap_final:.4f}")

if loss_gap_final > 0.1:
    print("⚠️  WARNING: Potential overfitting detected (val_loss >> train_loss)")
elif loss_gap_final < -0.05:
    print("⚠️  NOTE: Model may benefit from more training")
else:
    print("✓ Good generalization (train and val losses are close)")

# ============================================================================
# SECTION 9: Model Evaluation on Test Set
# ============================================================================

print("\n" + "="*80)
print("EVALUATING ON TEST SET")
print("="*80)

# Predict on test set
predictions = trainer.predict(test_dataset)
pred_labels = np.argmax(predictions.predictions, axis=1)
true_labels = test_df['label_encoded'].values

# Calculate metrics
test_accuracy = accuracy_score(true_labels, pred_labels)
test_f1 = f1_score(true_labels, pred_labels, average='weighted')

print(f"\n✓ Test Set Performance:")
print(f"  Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"  F1 Score: {test_f1:.4f}")

# Detailed classification report
print("\n" + "-"*80)
print("CLASSIFICATION REPORT")
print("-"*80)
report = classification_report(
    true_labels, pred_labels,
    target_names=[reverse_mapping[i] for i in sorted(reverse_mapping.keys())],
    digits=4
)
print(report)

# Confusion Matrix
cm = confusion_matrix(true_labels, pred_labels)

# Normalize confusion matrix
cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

# Plot confusion matrices
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Raw counts
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=[reverse_mapping[i] for i in sorted(reverse_mapping.keys())],
            yticklabels=[reverse_mapping[i] for i in sorted(reverse_mapping.keys())])
axes[0].set_title('Confusion Matrix (Counts)', fontsize=14, fontweight='bold')
axes[0].set_ylabel('True Label', fontsize=12)
axes[0].set_xlabel('Predicted Label', fontsize=12)

# Normalized
sns.heatmap(cm_normalized, annot=True, fmt='.2%', cmap='Greens', ax=axes[1],
            xticklabels=[reverse_mapping[i] for i in sorted(reverse_mapping.keys())],
            yticklabels=[reverse_mapping[i] for i in sorted(reverse_mapping.keys())])
axes[1].set_title('Confusion Matrix (Normalized)', fontsize=14, fontweight='bold')
axes[1].set_ylabel('True Label', fontsize=12)
axes[1].set_xlabel('Predicted Label', fontsize=12)

plt.tight_layout()
plt.show()

# ============================================================================
# SECTION 10: Per-Class Performance Analysis
# ============================================================================

print("\n" + "="*80)
print("PER-CLASS PERFORMANCE ANALYSIS")
print("="*80)

# Calculate per-class metrics
per_class_metrics = []
for i in sorted(reverse_mapping.keys()):
    true_binary = (true_labels == i).astype(int)
    pred_binary = (pred_labels == i).astype(int)

    tp = np.sum((true_binary == 1) & (pred_binary == 1))
    fp = np.sum((true_binary == 0) & (pred_binary == 1))
    fn = np.sum((true_binary == 1) & (pred_binary == 0))
    tn = np.sum((true_binary == 0) & (pred_binary == 0))

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    per_class_metrics.append({
        'Class': reverse_mapping[i],
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1,
        'Support': np.sum(true_binary)
    })

metrics_df = pd.DataFrame(per_class_metrics)
print("\n", metrics_df.to_string(index=False))

# Visualize per-class metrics
fig = go.Figure()
fig.add_trace(go.Bar(x=metrics_df['Class'], y=metrics_df['Precision'],
                     name='Precision', marker_color='lightblue'))
fig.add_trace(go.Bar(x=metrics_df['Class'], y=metrics_df['Recall'],
                     name='Recall', marker_color='lightcoral'))
fig.add_trace(go.Bar(x=metrics_df['Class'], y=metrics_df['F1-Score'],
                     name='F1-Score', marker_color='lightgreen'))

fig.update_layout(
    title='Per-Class Performance Metrics',
    xaxis_title='CEFR Level',
    yaxis_title='Score',
    barmode='group',
    height=500
)
fig.show()

# ============================================================================
# SECTION 11: Error Analysis
# ============================================================================

print("\n" + "="*80)
print("ERROR ANALYSIS")
print("="*80)

# Find misclassified examples
misclassified_idx = np.where(pred_labels != true_labels)[0]
print(f"\nTotal misclassified: {len(misclassified_idx)} out of {len(test_df)}")
print(f"Error rate: {len(misclassified_idx)/len(test_df)*100:.2f}%")

# Analyze error patterns
error_analysis = []
for idx in misclassified_idx[:20]:  # Show first 20 errors
    true_label = reverse_mapping[true_labels[idx]]
    pred_label = reverse_mapping[pred_labels[idx]]
    text = test_df.iloc[idx][text_col][:200]  # First 200 chars

    error_analysis.append({
        'True': true_label,
        'Predicted': pred_label,
        'Text Preview': text + '...'
    })

if error_analysis:
    error_df = pd.DataFrame(error_analysis)
    print("\nSample Misclassifications:")
    print(error_df.to_string(index=False))

# ============================================================================
# SECTION 12: Model Confidence Analysis
# ============================================================================

print("\n" + "="*80)
print("MODEL CONFIDENCE ANALYSIS")
print("="*80)

# Get prediction probabilities
probs = torch.softmax(torch.tensor(predictions.predictions), dim=1).numpy()
max_probs = np.max(probs, axis=1)

# Calculate average confidence
avg_confidence = np.mean(max_probs)
print(f"\nAverage prediction confidence: {avg_confidence:.4f}")

# Confidence vs Correctness
correct_mask = (pred_labels == true_labels)
correct_confidence = max_probs[correct_mask]
incorrect_confidence = max_probs[~correct_mask]

print(f"Average confidence (correct predictions): {np.mean(correct_confidence):.4f}")
print(f"Average confidence (incorrect predictions): {np.mean(incorrect_confidence):.4f}")

# Visualize confidence distribution
fig = go.Figure()
fig.add_trace(go.Histogram(x=correct_confidence, name='Correct',
                           opacity=0.7, marker_color='green', nbinsx=30))
fig.add_trace(go.Histogram(x=incorrect_confidence, name='Incorrect',
                           opacity=0.7, marker_color='red', nbinsx=30))

fig.update_layout(
    title='Prediction Confidence Distribution',
    xaxis_title='Confidence Score',
    yaxis_title='Count',
    barmode='overlay',
    height=400
)
fig.show()

# ============================================================================
# SECTION 13: Final Summary and Recommendations
# ============================================================================

print("\n" + "="*80)
print("FINAL SUMMARY")
print("="*80)

print(f"\n📊 Model Performance:")
print(f"   ✓ Test Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"   ✓ Test F1 Score: {test_f1:.4f}")
print(f"   ✓ Average Confidence: {avg_confidence:.4f}")

print(f"\n🎯 Target Achievement:")
target_accuracy = 0.90
if test_accuracy >= target_accuracy:
    print(f"   ✓ TARGET MET! Achieved ≥{target_accuracy*100}% accuracy")
else:
    gap = (target_accuracy - test_accuracy) * 100
    print(f"   ⚠️  {gap:.2f}% away from {target_accuracy*100}% target")
    print(f"\n💡 Recommendations to improve:")
    print(f"   • Increase training data (especially for underperforming classes)")
    print(f"   • Try BERT-base-uncased instead of DistilBERT")
    print(f"   • Adjust learning rate or training epochs")
    print(f"   • Apply data augmentation techniques")

print(f"\n🚫 Overfitting Check:")
if abs(loss_gap_final) < 0.1:
    print(f"   ✓ No significant overfitting detected")
else:
    print(f"   ⚠️  Review training curves above")

print(f"\n💾 Model saved to: ./best_model/")
print(f"   Use this model for deployment in your web platform")

print("\n" + "="*80)
print("TRAINING COMPLETE!")
print("="*80)
```
