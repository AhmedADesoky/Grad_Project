```python
import pandas as pd
import random
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet, stopwords
import os

nltk_data_dir = r"C:\Users\Ahmed Abdel Samad\AppData\Roaming\nltk_data"
os.makedirs(nltk_data_dir, exist_ok=True)
nltk.data.path.append(nltk_data_dir)

nltk.download('punkt', download_dir=nltk_data_dir, quiet=True)
nltk.download('punkt_tab', download_dir=nltk_data_dir, quiet=True)
nltk.download('stopwords', download_dir=nltk_data_dir, quiet=True)
nltk.download('wordnet', download_dir=nltk_data_dir, quiet=True)
nltk.download('omw-1.4', download_dir=nltk_data_dir, quiet=True)

stop_words = set(stopwords.words('english'))

DATA_PATH = r"C:\Grad_Project\Models\Datasets\final_combined_cefr_english_dataset.csv"
VOCAB1_PATH = r"C:\Grad_Project\Models\Datasets\cefrj-vocabulary-profile-1.5.csv"
VOCAB2_PATH = r"C:\Grad_Project\Models\Datasets\octanove-vocabulary-profile-c1c2-1.0.csv"

OUTPUT_AUG = r"C:\Grad_Project\Models\Datasets\FINAL_HIGH_QUALITY_AUGMENTED.csv"
OUTPUT_FINAL = r"C:\Grad_Project\Models\Datasets\FINAL_PERFECT_BALANCED_5000.csv"

df = pd.read_csv(DATA_PATH)
print("Original distribution:")
print(df['level'].value_counts().sort_index())

vocab1 = pd.read_csv(VOCAB1_PATH)[['headword','CEFR']]
vocab2 = pd.read_csv(VOCAB2_PATH)[['headword','CEFR']]

vocab = pd.concat([vocab1, vocab2]).dropna()
vocab['headword'] = vocab['headword'].str.lower().str.strip()
vocab['CEFR'] = vocab['CEFR'].str.upper().str.strip()

level_words = {}
for lvl in ['A1','A2','B1','B2','C1','C2']:
    level_words[lvl] = set(vocab[vocab['CEFR'] == lvl]['headword'])

print("Vocabulary sizes:", {k: len(v) for k, v in level_words.items()})

def get_best_synonyms(word, level):
    best_syns = []
    word_synsets = wordnet.synsets(word)
    if not word_synsets:
        return []
    
    main_synset = word_synsets[0]
    
    for syn in word_synsets:
        for lemma in syn.lemmas():
            s = lemma.name().lower().replace('_', ' ')
            if s != word and s in level_words[level]:
                s_synsets = wordnet.synsets(s)
                if s_synsets:
                    similarity = main_synset.wup_similarity(s_synsets[0])
                    if similarity and similarity > 0.75:  # Very high similarity
                        best_syns.append(s)
    
    return list(set(best_syns))  

def final_augment(text, level):
    tokens = word_tokenize(text)
    new_tokens = tokens.copy()
    
    candidates = []
    for i, token in enumerate(tokens):
        word = token.lower().rstrip('.,!?;:"()[]{}')
        punct = token[len(word):]
        if (word.isalpha() and 
            word not in stop_words and 
            len(word) > 3 and 
            word in level_words[level]):
            syns = get_best_synonyms(word, level)
            if syns:
                candidates.append((i, word, punct, token, syns))
    
    if len(candidates) < 2:
        return text
    
    num_replace = random.randint(2, min(5, len(candidates)))
    chosen = random.sample(candidates, num_replace)
    
    changed = False
    for i, old_word, punct, old_token, syns in chosen:
        new_word = random.choice(syns)
        if old_token[0].isupper():
            new_word = new_word.capitalize()
        new_tokens[i] = new_word + punct
        changed = True
    
    return ' '.join(new_tokens) if changed else text

TARGET = 5000
augmented = []
examples = []

print(f"\n{'='*80}")
print(f"FINAL HIGH-QUALITY SYNONYM AUGMENTATION → TARGET {TARGET} PER CLASS")
print(f"{'='*80}")

for level in ['A1','A2','B1','B2','C1','C2']:
    level_df = df[df['level'] == level]
    current = len(level_df)
    needed = TARGET - current
    if needed <= 0:
        print(f"{level}: {current} (sufficient)")
        continue
    
    print(f"{level}: {current} → generating {needed:,} samples...")
    pool = level_df['text'].tolist()
    count = 0
    
    while count < needed:
        orig = random.choice(pool)
        aug = final_augment(orig, level)
        if aug != orig and len(aug) > 40:
            augmented.append({'text': aug, 'level': level})
            count += 1
            if len(examples) < 10:
                examples.append((orig, aug))
            if count % 1000 == 0:
                print(f"   → {count:,}/{needed:,}")
    
    print(f"{level}: COMPLETED ({count:,} generated)")

# Combine original + augmented
temp_df = pd.concat([df, pd.DataFrame(augmented)], ignore_index=True)

# Remove duplicates
temp_df.drop_duplicates(subset=['text'], inplace=True)

# === FINAL BALANCING: Force exactly 5000 per class ===
print("\n" + "="*80)
print("FINAL BALANCING: ENSURING EXACTLY 5000 SAMPLES PER CLASS")
print("="*80)

balanced_parts = []
for level in ['A1','A2','B1','B2','C1','C2']:
    level_subset = temp_df[temp_df['level'] == level]
    current_count = len(level_subset)
    print(f"{level}: {current_count:,} samples available")
    
    # Resample to exactly 5000 (oversample if <5000, downsample if >5000)
    resampled = level_subset.sample(n=5000, replace=True, random_state=42)
    balanced_parts.append(resampled)

final_df = pd.concat(balanced_parts, ignore_index=True)
final_df = final_df.sample(frac=1, random_state=42).reset_index(drop=True)  # Shuffle

# Save
final_df.to_csv(OUTPUT_FINAL, index=False)

print(f"\nFINAL PERFECTLY BALANCED DATASET SAVED: {OUTPUT_FINAL}")
print("Final distribution (exactly 5000 per class):")
print(final_df['level'].value_counts().sort_index())
print(f"Total samples: {len(final_df):,}")

print("\n" + " PERFECT HIGH-QUALITY EXAMPLES ".center(80, "="))
for i, (orig, aug) in enumerate(examples):
    print(f"\nExample {i+1}")
    print("Original :", orig[:400])
    print("Augmented:", aug[:400])
    print("-" * 80)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from transformers import BertTokenizer, BertModel, BertForMaskedLM
import pandas as pd
import numpy as np
from torch.utils.data import Dataset, DataLoader
import random
import os

# ============================================================================
# FAST HIGH-QUALITY NLP AUGMENTATION USING CONDITIONAL VAE-GAN + BERT
# ============================================================================

# Use the actual path where datasets are stored
BASE_PATH = r"C:\Grad_Project\Models\Datasets"
OUTPUT_PATH = os.path.join(BASE_PATH, "GAN_AUGMENTED_BALANCED.csv")

# Try multiple possible data sources (in order of preference)
POSSIBLE_DATA_FILES = [
    "PERFECTLY_BALANCED_CEFR_DATASET.csv",
    "FINAL_BALANCED_SUCCESS.csv", 
    "FINAL_PERFECT_BALANCED_DATASET.csv",
    "final_combined_cefr_english_dataset.csv",
]


def find_working_dataset():
    """Find the first working dataset file using robust CSV reading"""
    print("Searching for a working dataset...")
    for filename in POSSIBLE_DATA_FILES:
        filepath = os.path.join(BASE_PATH, filename)
        if os.path.exists(filepath):
            print(f"\nTrying: {filename}")
            try:
                # Method 1: Python engine with low_memory=False
                df = pd.read_csv(filepath, engine='python', low_memory=False)
                if 'text' in df.columns and 'level' in df.columns:
                    print(f"✓ Found working dataset: {filename}")
                    print(f"  Loaded {len(df)} rows successfully")
                    return filepath
                else:
                    print(f"✗ Missing 'text' or 'level' columns")
            except Exception as e1:
                # Method 2: Try with explicit dtype
                try:
                    df = pd.read_csv(filepath, dtype={'text': str, 'level': str}, low_memory=False)
                    if 'text' in df.columns and 'level' in df.columns:
                        print(f"✓ Found working dataset: {filename}")
                        print(f"  Loaded {len(df)} rows successfully (method 2)")
                        return filepath
                except Exception as e2:
                    print(f"✗ Failed to read: {str(e1)[:80]}")
    return None


class CEFRTextDataset(Dataset):
    """Dataset for CEFR-leveled texts"""
    def __init__(self, texts, levels, tokenizer, max_len=128):
        self.texts = texts
        self.levels = levels
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.level_map = {'A1': 0, 'A2': 1, 'B1': 2, 'B2': 3, 'C1': 4, 'C2': 5}
    
    def __len__(self):
        return len(self.texts)
    
    def __getitem__(self, idx):
        text = str(self.texts[idx])
        level = self.level_map[self.levels[idx]]
        
        encoding = self.tokenizer(
            text,
            max_length=self.max_len,
            padding='max_length',
            truncation=True,
            return_tensors='pt'
        )
        
        return {
            'input_ids': encoding['input_ids'].squeeze(),
            'attention_mask': encoding['attention_mask'].squeeze(),
            'level': torch.tensor(level, dtype=torch.long)
        }


class ConditionalTextEncoder(nn.Module):
    """Encoder: BERT + VAE encoding with level conditioning"""
    def __init__(self, bert_model, latent_dim=256, num_levels=6):
        super().__init__()
        self.bert = bert_model
        self.hidden_size = bert_model.config.hidden_size
        
        # Level embedding
        self.level_embed = nn.Embedding(num_levels, 64)
        
        # VAE components
        self.fc_mu = nn.Linear(self.hidden_size + 64, latent_dim)
        self.fc_logvar = nn.Linear(self.hidden_size + 64, latent_dim)
        
    def forward(self, input_ids, attention_mask, level):
        # Get BERT embeddings
        outputs = self.bert(input_ids, attention_mask=attention_mask)
        pooled = outputs.pooler_output  # [batch, 768]
        
        # Concatenate with level embedding
        level_emb = self.level_embed(level)  # [batch, 64]
        combined = torch.cat([pooled, level_emb], dim=1)
        
        # VAE encoding
        mu = self.fc_mu(combined)
        logvar = self.fc_logvar(combined)
        
        return mu, logvar


class ConditionalTextGenerator(nn.Module):
    """Generator: Latent → Token logits with level conditioning"""
    def __init__(self, latent_dim=256, vocab_size=30522, max_len=128, num_levels=6):
        super().__init__()
        self.latent_dim = latent_dim
        self.max_len = max_len
        
        self.level_embed = nn.Embedding(num_levels, 64)
        
        # Decoder network
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim + 64, 512),
            nn.LayerNorm(512),
            nn.GELU(),
            nn.Dropout(0.2),
            
            nn.Linear(512, 1024),
            nn.LayerNorm(1024),
            nn.GELU(),
            nn.Dropout(0.2),
            
            nn.Linear(1024, max_len * vocab_size)
        )
        
        self.vocab_size = vocab_size
        
    def forward(self, z, level):
        level_emb = self.level_embed(level)
        combined = torch.cat([z, level_emb], dim=1)
        
        logits = self.decoder(combined)
        logits = logits.view(-1, self.max_len, self.vocab_size)
        
        return logits


class ConditionalDiscriminator(nn.Module):
    """Discriminator: Real/Fake text classifier with level conditioning"""
    def __init__(self, bert_model, num_levels=6):
        super().__init__()
        self.bert = bert_model
        hidden_size = bert_model.config.hidden_size
        
        self.level_embed = nn.Embedding(num_levels, 64)
        
        self.classifier = nn.Sequential(
            nn.Linear(hidden_size + 64, 512),
            nn.LayerNorm(512),
            nn.LeakyReLU(0.2),
            nn.Dropout(0.3),
            
            nn.Linear(512, 256),
            nn.LayerNorm(256),
            nn.LeakyReLU(0.2),
            nn.Dropout(0.3),
            
            nn.Linear(256, 1),
            nn.Sigmoid()
        )
    
    def forward(self, input_ids, attention_mask, level):
        outputs = self.bert(input_ids, attention_mask=attention_mask)
        pooled = outputs.pooler_output
        
        level_emb = self.level_embed(level)
        combined = torch.cat([pooled, level_emb], dim=1)
        
        return self.classifier(combined)


class FastNLPAugmenter:
    """Complete augmentation system with training and generation"""
    def __init__(self, device='cuda' if torch.cuda.is_available() else 'cpu'):
        self.device = device
        print(f"Using device: {device}")
        self.tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
        
        # Initialize models
        print("Loading BERT models...")
        bert_encoder = BertModel.from_pretrained('bert-base-uncased')
        bert_discriminator = BertModel.from_pretrained('bert-base-uncased')
        
        self.encoder = ConditionalTextEncoder(bert_encoder).to(device)
        self.generator = ConditionalTextGenerator().to(device)
        self.discriminator = ConditionalDiscriminator(bert_discriminator).to(device)
        
        # For quick generation, also use BERT MLM
        self.mlm_model = BertForMaskedLM.from_pretrained('bert-base-uncased').to(device)
        
        # Optimizers
        self.opt_enc = optim.AdamW(self.encoder.parameters(), lr=2e-5)
        self.opt_gen = optim.AdamW(self.generator.parameters(), lr=1e-4)
        self.opt_disc = optim.AdamW(self.discriminator.parameters(), lr=1e-4)
        
    def reparameterize(self, mu, logvar):
        """VAE reparameterization trick"""
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    
    def train_step(self, batch):
        """Single training step for the VAE-GAN"""
        input_ids = batch['input_ids'].to(self.device)
        attention_mask = batch['attention_mask'].to(self.device)
        level = batch['level'].to(self.device)
        
        batch_size = input_ids.size(0)
        
        # ==========================
        # Train Discriminator
        # ==========================
        self.opt_disc.zero_grad()
        
        # Real samples
        real_pred = self.discriminator(input_ids, attention_mask, level)
        real_loss = nn.BCELoss()(real_pred, torch.ones_like(real_pred))
        
        # Fake samples
        mu, logvar = self.encoder(input_ids, attention_mask, level)
        z = self.reparameterize(mu, logvar)
        fake_logits = self.generator(z, level)
        fake_ids = torch.argmax(fake_logits, dim=-1)
        fake_mask = (fake_ids != 0).long()
        
        fake_pred = self.discriminator(fake_ids.detach(), fake_mask, level)
        fake_loss = nn.BCELoss()(fake_pred, torch.zeros_like(fake_pred))
        
        d_loss = (real_loss + fake_loss) / 2
        d_loss.backward()
        self.opt_disc.step()
        
        # ==========================
        # Train Generator + Encoder
        # ==========================
        self.opt_enc.zero_grad()
        self.opt_gen.zero_grad()
        
        mu, logvar = self.encoder(input_ids, attention_mask, level)
        z = self.reparameterize(mu, logvar)
        fake_logits = self.generator(z, level)
        
        # Reconstruction loss
        recon_loss = nn.CrossEntropyLoss()(
            fake_logits.reshape(-1, self.generator.vocab_size),
            input_ids.reshape(-1)
        )
        
        # KL divergence
        kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
        kl_loss = kl_loss / batch_size
        
        # GAN loss (fool discriminator)
        fake_ids = torch.argmax(fake_logits, dim=-1)
        fake_mask = (fake_ids != 0).long()
        gen_pred = self.discriminator(fake_ids, fake_mask, level)
        gan_loss = nn.BCELoss()(gen_pred, torch.ones_like(gen_pred))
        
        # Combined loss
        total_loss = recon_loss + 0.1 * kl_loss + 0.5 * gan_loss
        total_loss.backward()
        
        self.opt_enc.step()
        self.opt_gen.step()
        
        return {
            'd_loss': d_loss.item(),
            'recon_loss': recon_loss.item(),
            'kl_loss': kl_loss.item(),
            'gan_loss': gan_loss.item()
        }
    
    def generate_augmented_fast(self, text, level, num_variants=5):
        """
        FAST generation using BERT MLM + conditional sampling
        This is faster than full GAN generation for inference
        """
        self.mlm_model.eval()
        
        variants = []
        tokens = self.tokenizer.tokenize(text)
        
        if len(tokens) == 0:
            return [text] * num_variants
        
        for _ in range(num_variants):
            # Randomly mask 15-30% of tokens
            masked_tokens = tokens.copy()
            num_mask = random.randint(int(len(tokens) * 0.15), int(len(tokens) * 0.3))
            if num_mask == 0:
                num_mask = 1
            mask_indices = random.sample(range(len(tokens)), min(num_mask, len(tokens)))
            
            for idx in mask_indices:
                masked_tokens[idx] = '[MASK]'
            
            # Convert to input
            masked_text = ' '.join(masked_tokens)
            inputs = self.tokenizer(masked_text, return_tensors='pt', 
                                   max_length=128, truncation=True, padding=True)
            inputs = {k: v.to(self.device) for k, v in inputs.items()}
            
            # Generate predictions
            with torch.no_grad():
                outputs = self.mlm_model(**inputs)
                predictions = outputs.logits
            
            # Sample from top-k predictions
            predicted_ids = inputs['input_ids'].clone()
            mask_token_id = self.tokenizer.mask_token_id
            
            for i in range(predicted_ids.size(1)):
                if predicted_ids[0, i] == mask_token_id:
                    probs = torch.softmax(predictions[0, i], dim=-1)
                    # Top-k sampling
                    top_k = torch.topk(probs, k=10)
                    sampled_id = top_k.indices[torch.multinomial(top_k.values, 1)]
                    predicted_ids[0, i] = sampled_id
            
            # Decode
            generated_text = self.tokenizer.decode(predicted_ids[0], skip_special_tokens=True)
            variants.append(generated_text)
        
        return variants


# ============================================================================
# MAIN EXECUTION
# ============================================================================

print("="*80)
print("BERT VAE-GAN DATA AUGMENTER")
print("="*80)

# Find a working dataset
DATA_PATH = find_working_dataset()

if DATA_PATH is None:
    print("\n" + "="*80)
    print("ERROR: NO WORKING DATASET FOUND!")
    print("="*80)
    print("\nPlease run Cell 5 first to create 'final_combined_cefr_english_dataset.csv'")
    print("\nCell 5 combines the base datasets into a clean, unified format.")
    print("\nAfter running Cell 5, run this cell again.")
    print("="*80)
else:
    try:
        # Load the dataset with robust reading
        print(f"\nLoading dataset from: {os.path.basename(DATA_PATH)}")
        
        # Try python engine first
        try:
            df = pd.read_csv(DATA_PATH, engine='python', low_memory=False)
        except:
            # Fallback to explicit dtypes
            df = pd.read_csv(DATA_PATH, dtype={'text': str, 'level': str}, low_memory=False)
        
        # Clean the data
        df = df.dropna(subset=['text', 'level'])
        df['level'] = df['level'].astype(str).str.strip().str.upper()
        df = df[df['level'].isin(['A1', 'A2', 'B1', 'B2', 'C1', 'C2'])]
        
        print(f"Loaded {len(df)} samples")
        print("\nOriginal distribution:")
        print(df['level'].value_counts().sort_index())
        
        # Initialize augmenter
        print("\n" + "="*80)
        print("INITIALIZING VAE-GAN AUGMENTER")
        print("="*80)
        augmenter = FastNLPAugmenter()
        
        # Create dataset
        dataset = CEFRTextDataset(
            df['text'].tolist(),
            df['level'].tolist(),
            augmenter.tokenizer
        )
        
        dataloader = DataLoader(dataset, batch_size=16, shuffle=True)
        
        # Training loop (lightweight - just a few epochs)
        print("\n" + "="*80)
        print("TRAINING VAE-GAN (3 epochs)")
        print("="*80)
        num_epochs = 3
        
        for epoch in range(num_epochs):
            augmenter.encoder.train()
            augmenter.generator.train()
            augmenter.discriminator.train()
            
            total_losses = {'d_loss': 0, 'recon_loss': 0, 'kl_loss': 0, 'gan_loss': 0}
            
            for i, batch in enumerate(dataloader):
                losses = augmenter.train_step(batch)
                for k, v in losses.items():
                    total_losses[k] += v
                
                if i % 50 == 0:
                    print(f"Epoch {epoch+1}/{num_epochs}, Step {i}: " + 
                          ", ".join([f"{k}={v:.4f}" for k, v in losses.items()]))
            
            avg_losses = {k: v/len(dataloader) for k, v in total_losses.items()}
            print(f"\nEpoch {epoch+1} Average Losses: {avg_losses}")
        
        # Fast generation for augmentation
        print("\n" + "="*80)
        print("GENERATING AUGMENTED DATA")
        print("="*80)
        
        TARGET = 5000
        augmented_data = []
        
        for level in ['A1', 'A2', 'B1', 'B2', 'C1', 'C2']:
            level_df = df[df['level'] == level]
            current = len(level_df)
            needed = TARGET - current
            
            if needed <= 0:
                print(f"{level}: {current} samples (sufficient)")
                continue
            
            print(f"{level}: Generating {needed} samples...")
            pool = level_df['text'].tolist()
            count = 0
            
            for i in range(0, needed, 5):  # Generate 5 variants at a time
                orig_text = random.choice(pool)
                variants = augmenter.generate_augmented_fast(orig_text, level, num_variants=5)
                
                for variant in variants:
                    if count >= needed:
                        break
                    if len(variant) > 40 and variant != orig_text:
                        augmented_data.append({'text': variant, 'level': level})
                        count += 1
                
                if count % 500 == 0 and count > 0:
                    print(f"  → {count}/{needed}")
            
            print(f"{level}: COMPLETED ({count} generated)")
        
        # Combine and balance
        final_df = pd.concat([df, pd.DataFrame(augmented_data)], ignore_index=True)
        final_df = final_df.drop_duplicates(subset=['text'])
        
        # Balance to exactly 5000 per class
        balanced_parts = []
        for level in ['A1', 'A2', 'B1', 'B2', 'C1', 'C2']:
            level_subset = final_df[final_df['level'] == level]
            resampled = level_subset.sample(n=5000, replace=True, random_state=42)
            balanced_parts.append(resampled)
        
        final_df = pd.concat(balanced_parts, ignore_index=True)
        final_df = final_df.sample(frac=1, random_state=42).reset_index(drop=True)
        
        # Save
        final_df.to_csv(OUTPUT_PATH, index=False)
        
        print(f"\n" + "="*80)
        print(f"✓ SAVED: {os.path.basename(OUTPUT_PATH)}")
        print("="*80)
        print("Final distribution:")
        print(final_df['level'].value_counts().sort_index())
        print(f"Total: {len(final_df):,} samples")
        
        # Show examples
        print("\n" + "="*80)
        print("AUGMENTATION EXAMPLES")
        print("="*80)
        for i in range(3):
            orig = random.choice(df['text'].tolist())
            level = random.choice(['A1', 'A2', 'B1', 'B2', 'C1', 'C2'])
            variants = augmenter.generate_augmented_fast(orig, level, num_variants=2)
            
            print(f"\nOriginal ({level}):")
            print(orig[:200])
            print("\nGenerated variants:")
            for j, var in enumerate(variants):
                print(f"  {j+1}. {var[:200]}")
            print("-" * 80)
            
    except Exception as e:
        print(f"\n" + "="*80)
        print(f"ERROR: {e}")
        print("="*80)
        import traceback
        traceback.print_exc()
        print("\n💡 SOLUTION: Run Cell 5 to regenerate the base dataset.")
```
